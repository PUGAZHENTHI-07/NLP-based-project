```python
def classify_spam(message_text):
    message_text = message_text.lower()
    spam_keywords = [
        "congratulations", "won", "prize", "gift card", "claim now",
        "free", "winner", "cash", "urgent", "limited time",
        "click here", "deal", "guarantee", "opportunity", "investment",
        "$$$", "money back", "credit card", "account suspended",
        "verify your account", "lotto", "lottery", "dear friend",
        "unsubscribe", "http", "https", ".com", ".net", ".org", ".biz"
    ]

    for keyword in spam_keywords:
        if keyword in message_text:
            return "Spam"
    return "Not Spam"
```

```python
message1 = "Congratulations! You have won a $1000 gift card. Click the link now to claim."
print(classify_spam(message1))

message2 = "Hi, just wanted to confirm our meeting for tomorrow at 10 AM."
print(classify_spam(message2))

message3 = "Your account has been suspended due to unusual activity. Please verify your details here: http://suspicious-link.com"
print(classify_spam(message3))
```
